## Aplique seu conhecimento 2.2 (p.40) - Despesas com cuidados da saúde

A base abaixo mostra as despesas anuais, em 2015, com cuidados da saúde por pessoa, nos 35 países com os maiores valores do PIB em 2015.

```python
cuidasaude = pd.read_excel(f'ex02-02health.xls')
```

```python
cuidasaude.sort_values('Dollars',ascending=False)
```

Os Estados Unidos com 9.536 dólares internacionais por pessoa, é um alto valor atípico superior. Ache a média das despesas com cuidados da saúde nesses países, com e sem os Estados Unidos. De quanto o valor atípico aumenta a média?

```python
print(cuidasaude['Dollars'].mean())
```

```python
print(cuidasaude[cuidasaude.Country!="UnitedStates"]['Dollars'].mean())
```

Ao retirar os Estados Unidos, podemos observar que a média teve uma redução de mais de 200 dólares. Será que a mediana sofre também uma grande alteração com a retirada desse valor?

```python
print(cuidasaude['Dollars'].median())
```

```python
print(cuidasaude[cuidasaude.Country!="UnitedStates"]['Dollars'].median())
```

Podemos observar que a mediana não sofreu tanta alteração, pois é resistente a valores atípicos.

Vamos visualizar o gráfico para verificar quão distante esse possível valor atípico está dos outros países.

```python
alt.Chart(cuidasaude).mark_bar().encode(
    alt.X("Dollars:Q", bin=True),
    y='count()',
)
```

O Altair está considerando intervalos de tamanho 1000. Como ficaria o gráfico se quiséssemos diminuir o tamanho dos intervalos das classes?

```python
alt.Chart(cuidasaude).mark_bar().encode(
    alt.X("Dollars:Q", bin=alt.Bin(step=200)),
    y='count()',
)
```

Ficaria muito mais difícil de perceber o padrão geral dessa distribuição. Apesar de não ter uma escolha certa para os tamanhos dos intervalos, essa escolha é determinante para gerar uma boa visualização da distribuição de uma variável quantitativa.

# Aula 9 - Quartis, Valores Atípicos e Boxplot

## Exemplo 2.4 (p.42) - Tempo de viagem para o trabalho Carolina do Norte

Voltando aos dados de Carolina do Norte, vamos analisar o intervalo interquartil como uma medida de dispersão.

Podemos calcular todas as medidas descritivas de uma só vez, com a função describe.

```python
tempocn.describe()
```

Dessa forma, temos que o Q1 é 13.5 e Q3 é 35. Podemos, portando, calcular a amplitude interquartil como a diferença desses dois quartis.

```python
aiq_cn = 35-13.5
print(f"AIQ = {aiq_cn}")
```

Para avaliarmos se 70 é de fato um potencial valor atípico, como supusemos acima ao avaliamos o tanto que esse valor influencia na média, vamos usar a regra 1.5xAIQ.

Para um valor ser considerado potencialmente atípico, ele precisa estar fora do intervalor que [Q1 - 1.5xAIQ; Q3 + 1.5xAIQ], que nesse caso é o intervalo:

```python
print(f"[{3.5-1.5*aiq_cn}; {35+1.5*(aiq_cn)}]")
```

Confirmamos, portanto, que por essa regra, o tempo de 70 minutos pode ser considerado um valor atípico.

Vamos visualizar o boxplot dessa distribuição.

```python
alt.Chart(tempocn).mark_boxplot(extent='min-max').encode(
    y='Minutes:Q'
).properties(width=200)
```

Esse boxplot do ponto mínimo ao ponto máximo da distribuição, o que foi determinado pelo parâmetro `extent="min-max`. Para considerarmos a visualização que destaque o valor atípico, pela regra 1.5xAIQ, podemos alterar esse parâmetro para `extent=1.5`.

```python
alt.Chart(tempocn).mark_boxplot(extent=1.5).encode(
    y='Minutes:Q'
).properties(width=200)
```

## Os dois juntos considerando valor atípico

Para fazer uma análise comparativa das distribuições, vamos visualizar os dois boxplots juntos em um mesmo gráfico.

Para isso, é preciso criar uma base que agregue os dois dados.

```python
tempocn['Origin'] = "Carolina do Norte"
tempony['Origin'] = "Nova York"

tempo_viagem = pd.concat([tempocn,tempony])

tempo_viagem
```

```python
alt.Chart(tempo_viagem).mark_boxplot(extent=1.5, size=40).encode(
    alt.X("Minutes:Q"),
    alt.Y("Origin:N")
    #alt.Color("Origin:N")
).properties(height=200)
```

Essa análise comparativa mostra a assimetria à direita mais acentuada para Carolina do Norte, inclusive com a presença de um valor atípico.

Mas deixa claro também, pelo tamanho do AIQ, que a dispersão da distribuição de Nova York é maior. O tamanho do boxplot de NY e a falta de valores atípicos também aponta para uma distribuição que se estende por um intervalo maior.

Além disso, no geral, o tempo de viagem dos moradores de NY até o trabalho é maior do que dos moradores de Carolina do Norte, já que há um deslocamento do gráfico todo de Nova York para a direita.

# Aula 10 - Desvio Padrão

## Exemplo das notas das turmas

```python
turma_a = [4,5,5,6,6,7,7,8]
turma_b = [1,2,4,6,6,9,10,10]
turma_c = [0,6,7,7,7,7.5,7.5]
```

```python
from statistics import stdev, mean, median

print(f"Turma A: média = {mean(turma_a)} ; desvio padrão = {stdev(turma_a)}")
print(f"Turma B: média = {mean(turma_b)} ; desvio padrão = {stdev(turma_b)}")
print(f"Turma C: média = {mean(turma_c)} ; desvio padrão = {stdev(turma_c)}")
```

```python
# outra forma de calcular, com describe
print(pd.DataFrame(turma_a).describe().loc[['mean','std']])
print(pd.DataFrame(turma_b).describe().loc[['mean','std']])
print(pd.DataFrame(turma_c).describe().loc[['mean','std']])
```

Verificamos, através das médias, que os alunos das três turmas tenderam a ter as notas em tomo de seis, mas, pelos desvios padrões, concluímos que os alunos da Turma A obtiveram notas relativamente próximas umas das outras, quando comparados aos alunos das outras turmas. Por outro lado, as notas dos alunos da Turma B foram as que se apresentaram mais heterogêneas.
