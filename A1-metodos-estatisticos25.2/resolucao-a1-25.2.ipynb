# Análise Univariada — 1 variável

Você olha uma coluna só do seu DataFrame, isoladamente, e tenta descrever como ela se distribui.

A pergunta que você está respondendo: "como são os valores dessa variável? Onde eles se concentram? O quanto variam? Tem algo estranho (outlier)?"

O que muda na prática:

Você usa uma coluna só pra fazer o gráfico (df["coluna"])
Gráfico depende do tipo da variável:
Quantitativa → histograma ou boxplot
Categórica → gráfico de barras (contagem ou % por categoria)
Estatísticas que você calcula: média, mediana, moda, desvio padrão, quartis, IQR, resumo dos 5 números — tudo isso é sobre essa única variável.
Sua análise final descreve: forma (simétrica/assimétrica), centro (média ou mediana) e dispersão (desvio padrão ou IQR) — exatamente o roteiro que já vimos.

# Análise Bivariada — 2 variáveis

Você olha duas colunas ao mesmo tempo e tenta entender se e como elas se relacionam.

A pergunta que você está respondendo: "quando X muda, Y também muda? Numa direção específica? A relação é forte ou fraca?"

Aqui você sempre define qual é a variável explicativa (independente, a possível "causa") e qual é a variável resposta (dependente, o que você quer entender) — lembra do exemplo das latas de cerveja → álcool no sangue?

O que muda na prática — depende da combinação de tipos das duas variáveis:

Combinação	Gráfico	O que você calcula

Categórica × Categórica	Tabela de contingência (+ gráfico de barras empilhadas/agrupadas)	% da resposta dentro de cada categoria da explicativa

Quantitativa × Quantitativa	Diagrama de dispersão (scatter plot)	Coeficiente de Pearson (r), direção, forma, intensidade

Categórica × Quantitativa	Boxplots lado a lado (um boxplot da quantitativa, para cada categoria)	Comparar centro/dispersão da quantitativa entre os grupos

Sua análise final descreve: direção (positiva/negativa), forma (linear/curva), intensidade (forte/fraca), e sempre com a ressalva de que correlação não implica causalidade.
