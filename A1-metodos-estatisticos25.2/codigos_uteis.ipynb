```python
minimo = microdados_enem["TP_FAIXA_ETARIA"].min()
q1 = microdados_enem["TP_FAIXA_ETARIA"].quantile(0.25)
mediana = microdados_enem["TP_FAIXA_ETARIA"].median()
q3 = microdados_enem["TP_FAIXA_ETARIA"].quantile(0.75)
maximo = microdados_enem["TP_FAIXA_ETARIA"].max()
print("Mínimo:", minimo)
print("Q1:", q1)
print("Mediana:", mediana)
print("Q3:", q3)
print("Máximo:", maximo)

média = microdados_enem["TP_FAIXA_ETARIA"].mean()
desvio_padrão = microdados_enem["TP_FAIXA_ETARIA"].std ()
print("Média:", média)
print("Desvio padrão:", desvio_padrão)

iqr = q3 - q1
print("IQR:", iqr)

limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

print("Limite inferior:", limite_inferior)
print("Limite superior:", limite_superior)

outliers = microdados_enem[(microdados_enem["TP_FAIXA_ETARIA"] < limite_inferior) | (microdados_enem["TP_FAIXA_ETARIA"] > limite_superior)]
print(outliers['Unnamed: 0'])
```
